```
import csv

input_tsv = "dataset/line_index.tsv"
output_csv = "dataset/metadata-new.csv"

with open(input_tsv, "r", encoding="utf-8") as fin, \
     open(output_csv, "w", encoding="utf-8", newline="") as fout:

    writer = csv.writer(fout, delimiter="|")

    for line in fin:
        line = line.strip()
        if not line:
            continue

        audio_id, text = line.split("\t", 1)

        # LJSpeech format:
        # filename|raw_text|normalized_text
        writer.writerow([audio_id, text, text])

print("metadata.csv created.")
```
